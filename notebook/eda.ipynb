```python
import fiftyone as fo
import fiftyone.brain as fob
from pathlib import Path
import json

# Load metadata img path + label
with open("./../data/module_metadata.json", "r") as f:
    data = json.load(f)

samples = []
base_dir = Path("./../data")

for item in data.values():
    img_path = base_dir / item["image_filepath"]
    sample = fo.Sample(filepath=img_path)
    sample["ground_truth"] = fo.Classification(label=item["anomaly_class"])
    samples.append(sample)

if fo.dataset_exists("InfraredSolar"):
    fo.delete_dataset("InfraredSolar")

dataset = fo.Dataset("InfraredSolar")
dataset.add_samples(samples)
```

```python

fob.compute_visualization(dataset, brain_key="image_embeddings", method="umap",)

fob.compute_uniqueness(dataset, uniqueness_field="uniqueness")
unique_view = dataset.sort_by("uniqueness", reverse=True).limit(50)

fob.compute_similarity(dataset, brain_key="image_similarity", similarity_field="similarity")
index = dataset.load_brain_results("image_similarity").find_duplicates(thresh=0.02)
```

```python
import fiftyone as fo
import imagehash

from PIL import Image
from collections import defaultdict

# 1. Calcul du pHash et regroupement
groups = defaultdict(list)

for sid, path in zip(*dataset.values(["id", "filepath"])):
    with Image.open(path) as img:
        h = str(imagehash.phash(img))
    groups[h].append(sid)

# 2. Garder uniquement les groupes avec plusieurs images
dups = [ids for ids in groups.values() if len(ids) > 1]

print(f"{len(dups)} groups of duplicates")
print(f"{sum(len(g) for g in dups)} images concerned")

# 3. Ajouter un numéro de groupe à chaque image
duplicate_ids = []

for group_id, ids in enumerate(dups):
    duplicate_ids.extend(ids)

    dataset.select(ids).set_values(
        "duplicate_group",
        [group_id] * len(ids)
    )

dataset.save()

# 4. Créer une vue contenant uniquement les doublons
dup_view = dataset.select(duplicate_ids)

# 5. Trier par groupe de doublons
dup_view = dup_view.sort_by("duplicate_group")

# 6. Ouvrir FiftyOne
session = fo.launch_app(dup_view)
```

```python
groups = defaultdict(list)

# Calcul du pHash
for sid, path in zip(*dataset.values(["id", "filepath"])):
    with Image.open(path) as img:
        h = str(imagehash.phash(img))
    groups[h].append(sid)

# Créer un groupe uniquement pour les groupes ayant plusieurs images
group_id = 0

for ids in groups.values():

    if len(ids) < 2:
        continue

    for sid in ids:
        dataset[sid]["similarity_group"] = group_id
        dataset[sid].save()

    group_id += 1

dataset.save()

print(f"{group_id} groups created")
```

```python
hashes = {}

for sid, path in zip(*dataset.values(["id", "filepath"])):
    with Image.open(path) as img:
        hashes[sid] = imagehash.phash(img)

# Chercher les images proches
similar_pairs = []

ids = list(hashes.keys())

for i in range(len(ids)):
    for j in range(i + 1, len(ids)):
        distance = hashes[ids[i]] - hashes[ids[j]]

        if distance <= 5:
            similar_pairs.append((ids[i], ids[j], distance))

print(f"{len(similar_pairs)} similar pairs found")
```

```python
import networkx as nx

G = nx.Graph()

for id1, id2, distance in similar_pairs:
    G.add_edge(id1, id2, distance=distance)

groups = list(nx.connected_components(G))

print(f"Number of groups : {len(groups)}")

for i, group in enumerate(groups[:20]):
    print(f"Group {i} : {len(group)} images")
```

```python
for group_id, group in enumerate(groups):

    for sid in group:
        dataset[sid]["similarity_group"] = group_id

dataset.save()
```

```python
total_grouped = sum(len(group) for group in groups)

print(f"Images belonging to a group : {total_grouped}")
print(f"Percentage of the dataset : {100 * total_grouped / len(dataset):.2f}%")
```

```python
for group_id, group in enumerate(groups):

    tag = f"similarity_group_{group_id}"

    dataset.select(list(group)).tag_samples(tag)

dataset.save()
```

```python
tags = set()

for sample in dataset:
    for tag in sample.tags:
        if tag.startswith("similarity_group_"):
            tags.add(tag)

print(f"{len(tags)} tag groups")
```

```python
from collections import Counter

for group_id, group in enumerate(groups):
    labels = [
        dataset[sid]["ground_truth"].label
        for sid in group
        if dataset[sid]["ground_truth"] is not None
    ]

    counts = Counter(labels)

    print(
        f"Group {group_id:03d} | "
        f"{len(group)} images | "
        f"{dict(counts)}"
    )
```
